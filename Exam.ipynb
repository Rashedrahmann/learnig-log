```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
iris = sns.load_dataset('iris')

# Display the first few rows just to see what we are working with
print(iris.head())
```

```python
# Create a histogram for every numerical column
iris.hist(figsize=(10, 8), bins=15, color='skyblue', edgecolor='black')

# Adjust layout for neatness and display
plt.suptitle('Histograms of Numerical Features')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

# We drop the 'species' column here because it's text, and box plots need numbers
sns.boxplot(data=iris.drop(columns='species'), palette='Set2')

plt.title('Box Plot of Iris Features')
plt.show()
```

```python
plt.figure(figsize=(6, 5))

# numeric_only=True ensures we don't get an error trying to correlate the text 'species' column
correlation_matrix = iris.corr(numeric_only=True)

# annot=True puts the actual numbers inside the colored boxes
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8, 6))

# 'hue' is the magic word in Seaborn — it automatically color-codes the dots based on the category
sns.scatterplot(data=iris, x='petal_length', y='petal_width', hue='species', s=100)

plt.title('Scatter Plot: Petal Length vs. Petal Width')
plt.show()
```

```python
# We use hue='species' again to see how the different flowers group together across all metrics
sns.pairplot(iris, hue='species', palette='husl')
plt.show()
```
